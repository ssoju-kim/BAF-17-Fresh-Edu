## (1) 이론
- 지난 세션시간까지 배운 데이터 분석 개념 점검을 위한 객관식/단답형 문제입니다.
- 코드셀에 정답을 적어주시면 됩니다.

#### 1. EDA(탐색적 데이터 분석, Exploratory Data Analysis)의 목적은?
① 데이터를 정제하는 것<br>
② 데이터를 시각화하고 특징을 파악하는 것<br>
③ 모델을 학습시키는 것<br>
④ 데이터를 수집하는 것<br>

정답: ② 데이터를 시각화하고 특징을 파악하는 것

#### 2. 다음 중 범주형 데이터를 수치형 데이터로 변환하는 방법이 아닌 것은?
① Label Encoding<br>
② One-Hot Encoding<br>
③ Min-Max Scaling<br>
④ Ordinal Encoding<br>

정답: ③ Min-Max Scaling

#### 3. 회귀 분석에서 사용하는 평가 지표로 적절하지 않은 것은?
① RMSE (Root Mean Squared Error)<br>
② R-squared<br>
③ AUC-ROC<br>
④ MAE (Mean Absolute Error)<br>

정답: ③ AUC-ROC

#### 4. 비지도 학습에 해당하는 알고리즘을 모두 고르시오.
① K-Means<br>
② PCA<br>
③ Random Forest<br>
④ DBSCAN<br>

정답: ① K-Means, ② PCA, ④ DBSCAN

#### 5. 분류 모델의 성능을 평가할 때 Precision과 Recall의 차이점은?

Precision이 중요한 상황과 Recall이 중요한 상황을 각각 적어주세요.

정답: Precision(정밀도)는 모델이 Positive로 예측한 것 중 실제 Positive의 비율이며 False Positive 비용이 클 때(ex. 광고 추천 시스템에서 잘못된 추천(불필요한 광고)을 최소화해야 할 때) 중요한 지표이고
Recall(재현율)은 실제 Positive 중 모델이 Positive로 잡아낸 비율이며 False Negative 비용이 클 때(ex. 사기 거래 탐지에서 가능한 모든 사기 거래를 찾아내야 할 때) 중요한 지표이다.

#### 6. 결측치 처리 방법에는 여러 가지가 있습니다. 아래와 같은 상황에서는 어떤 결측치 처리 방법이 가장 적절한가요?

- 데이터에서 일부 숫자형 변수의 결측치 비율이 40% 이상이다.
- 데이터를 분석해보니 결측값이 특정한 패턴 없이 무작위로 발생했다.

정답: 이 경우에는 결측치가 40% 이상으로 매우 많고 그 발생이 완전 무작위이기 때문에 평균·중앙값 대체나 KNN·회귀 대체 같은 보간방법을 쓰면 전체 분포가 크게 왜곡되고 불확실성도 커진다. 반면에 해당 변수를 아예 삭제하면 정보 손실은 있지만 모델에 잘못된 신호를 주는 위험을 막을 수 있고 분석 결과의 안정성이 높아진다. 따라서 40% 이상 결측이 있는 숫자형 변수는 과감히 제거하고 나머지 결측치 비율이 낮은 변수들에 대해서는 중앙값 대체나 간단한 imputer를 적용하는 것이 가장 합리적이다.

#### 7. K-means 알고리즘의 단점 중 아래와 같은 상황에서 어떤 방법을 적용할 수 있을까요?

거리 기반 알고리즘으로 속성(피쳐)의 개수가 많은 경우 정확도가 떨어졌다.

정답: K-means는 피처가 많아질수록 모든 샘플 간 유클리드 거리가 비슷해지는 ‘차원의 저주’ 현상으로 클러스터 경계가 모호해져 정확도가 떨어지는데 이를 해결하려면 PCA와 같은 차원 축소 기법으로 분산이 큰 주성분만 추출해 수십 차원 이하로 축소한 뒤 K-means를 적용하면 고차원 노이즈가 제거되어 더 명확하고 정확한 군집 결과를 얻을 수 있다.

#### 8. 앙상블 학습(Ensemble Learning) 방법 중 Bagging과 Boosting의 차이점

각각 대표적인 알고리즘을 하나씩 제시해주세요.

정답: Bagging은 원본 데이터를 복원 추출해 병렬로 다수의 약한 학습기를 학습시키고 예측을 평균·투표해 분산을 줄이는 기법으로 Random Forest가 대표적이며 Boosting은 이전 학습기의 오분류 샘플에 가중치를 부여해 순차적으로 학습기를 학습시켜 편향을 줄이고 예측 성능을 높이는 기법으로 AdaBoost가 대표적이다.

#### 9. 다중공선성이란 무엇이며, 다중공선성이 심할 경우 해결 방법은 무엇인가요?



정답: 다중공선성이란 회귀모형에서 독립변수들 간의 상호 상관관계가 매우 높아 회귀계수 추정치의 분산이 커지고 불안정해지는 현상을 말하며 다중공선성이 심할 경우에는 상관이 높은 변수를 제거하거나 변수 결합·차원 축소를 위해 PCA 기반 주성분회귀를 수행하거나 Ridge·Lasso 회귀와 같은 규제 기법을 적용해 계수의 분산을 줄이거나 VIF(분산팽창계수)를 기준으로 변수를 선별하는 방법 등을 적용할 수 있다.

#### 10. 머신러닝 모델의 과적합(Overfitting)을 방지하기 위한 방법을 2가지 이상 적어주세요.

머신러닝 모델의 과적합을 방지하기 위해서는 L1/L2 규제를 적용해 가중치 크기를 제한하거나 검증 손실이 증가하기 전에 학습을 멈추는 조기 종료와 교차 검증로 일반화 성능을 지속 확인하며 드롭아웃 또는 데이터 증강으로 학습 데이터를 다양화하는 등의 방법을 적용할 수 있다.
